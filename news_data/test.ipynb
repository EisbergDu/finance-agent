```python
import os
import json
import requests
import time
from urllib.parse import urlencode
from datetime import datetime
from pathlib import Path

API_BASE = "https://stocknewsapi.com/api/v1"
TOKEN = os.getenv("STOCKNEWS_TOKEN", "ea8w3dkncuzby3rot9oreutb5no33phj3uxzci9m")

# 目标保存目录（Windows 路径）
SAVE_DIR = Path(r"I:\\finance-agent\\news_data\\data\\stock")

def build_url(
    tickers="NVDA",
    items=50,
    page=1,
    date_range="01012024-10312025",
    time_range="000000-235959",
    search=None,
    source=None
):
    params = {
        "tickers": tickers,
        "items": str(items),
        "page": str(page),
        "date": date_range,
        "time": time_range,
        "token": TOKEN
    }
    if search:
        params["search"] = search
    if source:
        params["source"] = source
    return f"{API_BASE}?{urlencode(params)}"

def fetch_news(url, timeout=30, max_retries=3):
    """
    带重试机制的请求函数
    """
    for attempt in range(max_retries):
        try:
            r = requests.get(url, timeout=timeout, verify=True)
            r.raise_for_status()
            return r.json()
        except requests.exceptions.SSLError as e:
            print(f"    ⚠️  SSL错误 (尝试 {attempt + 1}/{max_retries}): {e}")
            if attempt < max_retries - 1:
                wait_time = (attempt + 1) * 5  # 递增等待时间：5秒、10秒、15秒
                print(f"    ⏳ 等待 {wait_time} 秒后重试...")
                time.sleep(wait_time)
            else:
                raise
        except requests.exceptions.ConnectionError as e:
            print(f"    ⚠️  连接错误 (尝试 {attempt + 1}/{max_retries}): {e}")
            if attempt < max_retries - 1:
                wait_time = (attempt + 1) * 5
                print(f"    ⏳ 等待 {wait_time} 秒后重试...")
                time.sleep(wait_time)
            else:
                raise
        except requests.exceptions.Timeout as e:
            print(f"    ⚠️  请求超时 (尝试 {attempt + 1}/{max_retries}): {e}")
            if attempt < max_retries - 1:
                wait_time = (attempt + 1) * 5
                print(f"    ⏳ 等待 {wait_time} 秒后重试...")
                time.sleep(wait_time)
            else:
                raise

def ensure_dir(path: Path):
    path.mkdir(parents=True, exist_ok=True)

def save_json(payload: dict, base_dir: Path, filename: str = "nvda_news_2024-2025.json") -> Path:
    """
    保存到固定文件名，每次覆盖
    """
    ensure_dir(base_dir)
    fp = base_dir / filename
    with fp.open("w", encoding="utf-8") as f:
        json.dump(payload, f, ensure_ascii=False, indent=2)
    return fp

def fetch_all_news(
    tickers="NVDA",
    date_range="01012024-10312025",
    items_per_page=100,
    delay=2,
    save_interval=5,
    start_page=1,
    output_filename="nvda_news_2024-2025.json"
):
    """
    自动翻页获取所有新闻
    
    Args:
        tickers: 股票代码
        date_range: 日期范围
        items_per_page: 每页获取数量
        delay: 每次请求间隔（秒）
        save_interval: 每多少页保存一次
        start_page: 从第几页开始（用于断点续传）
        output_filename: 输出文件名（固定）
    """
    all_news = []
    page = start_page
    consecutive_errors = 0  # 连续错误计数
    max_consecutive_errors = 3  # 最大连续错误次数
    
    print(f"=== 开始获取 {tickers} 新闻 ({date_range}) ===")
    print(f"从第 {start_page} 页开始")
    print(f"保存文件: {output_filename}\n")
    print("💡 提示: 按 Ctrl+C 可以随时中断并保存当前数据\n")
    
    try:
        while True:
            print(f"正在获取第 {page} 页...")
            
            url = build_url(
                tickers=tickers,
                items=items_per_page,
                page=page,
                date_range=date_range
            )
            
            try:
                data = fetch_news(url, timeout=30, max_retries=3)
                items = data.get("data") or []
                
                # 重置连续错误计数
                consecutive_errors = 0
                
                if not items:
                    print("  ✓ 没有更多数据了")
                    break
                
                all_news.extend(items)
                print(f"  ✓ 第 {page} 页获取 {len(items)} 条新闻 (总计: {len(all_news)})")
                
                # 定期保存（覆盖同一个文件）
                if page % save_interval == 0:
                    saved_path = save_json(
                        {"data": all_news, "last_page": page, "total": len(all_news)}, 
                        SAVE_DIR,
                        output_filename
                    )
                    print(f"  💾 已保存 {len(all_news)} 条新闻 (最后页码: {page})")
                
                # 如果返回的数量少于请求的数量，说明是最后一页
                if len(items) < items_per_page:
                    print("  ✓ 已到达最后一页")
                    break
                
                page += 1
                
                # 添加延迟避免触发速率限制
                print(f"  ⏳ 等待 {delay} 秒...")
                time.sleep(delay)
                    
            except requests.exceptions.HTTPError as e:
                consecutive_errors += 1
                if e.response.status_code == 429:
                    print(f"  ⚠️  触发速率限制 (429)，等待 60 秒后重试...")
                    save_json(
                        {"data": all_news, "last_page": page - 1, "total": len(all_news)}, 
                        SAVE_DIR,
                        output_filename
                    )
                    time.sleep(60)
                    consecutive_errors = 0  # 重置计数
                    continue
                else:
                    print(f"  ❌ HTTP错误 ({consecutive_errors}/{max_consecutive_errors}): {e}")
                    if consecutive_errors >= max_consecutive_errors:
                        print("  ❌ 连续错误次数过多，停止获取")
                        save_json(
                            {"data": all_news, "last_page": page - 1, "total": len(all_news)}, 
                            SAVE_DIR,
                            output_filename
                        )
                        break
                    time.sleep(10)
                    continue
                    
            except (requests.exceptions.SSLError, 
                    requests.exceptions.ConnectionError,
                    requests.exceptions.Timeout) as e:
                consecutive_errors += 1
                print(f"  ❌ 网络错误 ({consecutive_errors}/{max_consecutive_errors}): {type(e).__name__}")
                
                if consecutive_errors >= max_consecutive_errors:
                    print("  ❌ 连续错误次数过多，停止获取")
                    save_json(
                        {"data": all_news, "last_page": page - 1, "total": len(all_news)}, 
                        SAVE_DIR,
                        output_filename
                    )
                    break
                
                print(f"  ⏳ 等待 15 秒后继续...")
                time.sleep(15)
                continue
                
            except Exception as e:
                consecutive_errors += 1
                print(f"  ❌ 未知错误 ({consecutive_errors}/{max_consecutive_errors}): {e}")
                
                if consecutive_errors >= max_consecutive_errors:
                    print("  ❌ 连续错误次数过多，停止获取")
                    save_json(
                        {"data": all_news, "last_page": page - 1, "total": len(all_news)}, 
                        SAVE_DIR,
                        output_filename
                    )
                    break
                
                time.sleep(10)
                continue
                
    except KeyboardInterrupt:
        print("\n\n⚠️  用户中断程序")
        save_json(
            {"data": all_news, "last_page": page - 1, "total": len(all_news)}, 
            SAVE_DIR,
            output_filename
        )
    
    return all_news

if __name__ == "__main__":
    # 如果之前中断了，可以从指定页继续
    # start_page = 11  # 从第11页继续
    start_page = 1     # 从头开始
    
    # 固定的输出文件名
    output_file = "GLD_news_2024_2025.json"
    
    # 获取所有新闻
    all_news = fetch_all_news(
        tickers="GLD",
        date_range="01012024-10312025",
        items_per_page=100,
        delay=3,            # 增加到3秒，更稳定
        save_interval=5,    # 每5页保存一次
        start_page=start_page,
        output_filename=output_file
    )
    
    # 最终保存
    print("\n" + "="*60)
    print(f"✅ 完成！总共获取了 {len(all_news)} 条新闻")
    
    if all_news:
        saved_path = save_json(
            {"data": all_news, "total": len(all_news)}, 
            SAVE_DIR,
            output_file
        )
        print(f"📁 最终数据已保存到: {saved_path}")
        
        # 显示前5条示例
        print("\n前 5 条新闻示例:")
        for i, item in enumerate(all_news[:5], 1):
            title = item.get("title", "无标题")
            source = item.get("source", "未知来源")
            date = item.get("date", "无日期")
            print(f"{i}. [{source}] {title} | {date}")
    
    print("="*60)
```
